```python
#Task 14: Restaurant Ordering System
menu = {
    "Burger": 500,
    "Pizza": 1200,
    "Pasta": 800,
    "Drink": 200
}

for Key, value in menu.items():
    print(Key, ":", value)

orders = []

while True:
    item = input("Enter your item name:").title()

    if item in menu:
        quantity = int(input("Enter your item quantity:"))
        sub_total = menu[item] * quantity
        print("Item is available")

        order = {
            "item": item,
            "quantity": quantity,
            "sub_total": sub_total
        }

        orders.append(order)
        print(order)

    else:
        print("Item is not available")

    choice = input("Do you want more items? Yes/No: ")

    if choice == "No":
        break
total_bill=0
for order in orders:
    total_bill=total_bill+order["sub_total"]
tax=(total_bill*13)/100
final_bill=total_bill+tax

if total_bill >=3000:
    discount_per=(total_bill*10)/100
    final_bill=final_bill-discount_per
else:
    final_bill=total_bill+tax
def receipt():

    print("\n------Final Receipt------")
    for order in orders:
        print("\n Order :",order)
    print("\n Total Bill:",total_bill)
    if total_bill >= 3000:
        print("\nDiscount 10%:",discount_per)
    else:
        print("\n Tax 13%:",tax)
        print("\n Final Bill:",final_bill)
receipt()
```
