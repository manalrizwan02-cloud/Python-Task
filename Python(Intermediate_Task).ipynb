```python
# Task 14: Restaurant Ordering System

menu = {
    "Burger": 500,
    "Pizza": 1200,
    "Pasta": 800,
    "Drink": 200
}

for key, value in menu.items():
    print(key, ":", value)

orders = []

while True:

    item = input("Enter your item name: ").title()

    if item in menu:

        quantity = int(input("Enter your item quantity: "))

        sub_total = menu[item] * quantity

        print("Item is available")
        print("Sub Total:", sub_total)

        order = {
            "item": item,
            "quantity": quantity,
            "sub_total": sub_total
        }

        orders.append(order)

    else:
        print("Item is not available")

    choice = input("Do you want more items? Yes/No: ").title()

    if choice == "No":
        break

total_bill = 0

for order in orders:
    total_bill = total_bill + order["sub_total"]

tax = (total_bill * 13) / 100

if total_bill >= 3000:
    discount = (total_bill * 10) / 100
else:
    discount = 0


final_bill = total_bill + tax - discount

def receipt():

    print("\n------ Final Receipt ------")

    for order in orders:

        print("Item:", order["item"])
        print("Quantity:", order["quantity"])
        print("Sub Total:", order["sub_total"])
        print()

    print("Total Bill:", total_bill)
    print("Tax 13%:", tax)
    print("Discount:", discount)
    print("Final Bill:", final_bill)


receipt()
```

```python

# Task 15: Clinic Management System

patient = []


def add_appointment():
    id = int(input("Enter Patient ID: "))

    for patient_info in patient:
        if patient_info["ID"] == id:
            print("Patient ID already exists")
            return

    name = input("Enter Patient Name: ")
    age = int(input("Enter Patient Age: "))
    disease = input("Enter Disease: ")
    time = input("Enter Appointment Time: ")
    emergency = input("Emergency Appointment? Yes/No: ")

    patient_info = {
        "ID": id,
        "Name": name,
        "Age": age,
        "Disease": disease,
        "Time": time,
        "Emergency": emergency
    }

    patient.append(patient_info)
    print("Appointment Added Successfully")


def search_patient():
    id = int(input("Enter Patient ID: "))

    for patient_info in patient:
        if patient_info["ID"] == id:
            print("Patient Found")

            print("ID:", patient_info["ID"])
            print("Name:", patient_info["Name"])
            print("Age:", patient_info["Age"])
            print("Disease:", patient_info["Disease"])
            print("Time:", patient_info["Time"])
            print("Emergency:", patient_info["Emergency"])

            return

    print("Patient Not Found")


def display_appointments():

    for patient_info in patient:

        if patient_info["Emergency"] == "Yes":
            print("\nEmergency Patient:")
        else:
            print("\nNormal Patient:")

        print("ID:", patient_info["ID"])
        print("Name:", patient_info["Name"])
        print("Age:", patient_info["Age"])
        print("Disease:", patient_info["Disease"])
        print("Time:", patient_info["Time"])
        print("Emergency:", patient_info["Emergency"])


def cancel_appointment():
    id = int(input("Enter Patient ID: "))

    for patient_info in patient:
        if patient_info["ID"] == id:
            patient.remove(patient_info)
            print("Appointment Cancelled")
            return

    print("Patient Not Found")


def clinic_summary():
    emergency = 0
    normal = 0

    for patient_info in patient:
        if patient_info["Emergency"] == "Yes":
            emergency = emergency + 1
        else:
            normal = normal + 1

    print("\nClinic Summary")
    print("Total Patients:", len(patient))
    print("Emergency Patients:", emergency)
    print("Normal Patients:", normal)


while True:

    print("\n===== Clinic Appointment Manager =====")
    print("1. Add Appointment")
    print("2. Search Patient")
    print("3. Display Appointments")
    print("4. Cancel Appointment")
    print("5. Display Clinic Summary")
    print("6. Exit")

    choice = int(input("Enter Choice: "))

    if choice == 1:
        add_appointment()

    elif choice == 2:
        search_patient()

    elif choice == 3:
        display_appointments()

    elif choice == 4:
        cancel_appointment()

    elif choice == 5:
        clinic_summary()

    elif choice == 6:
        print("Thank You!")
        break

    else:
        print("Invalid Choice")
```
