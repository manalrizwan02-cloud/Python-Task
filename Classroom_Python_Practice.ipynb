```python
#Task 7
#Employee weekly salary
name=input("Enter Employee Name:")
T_workhours=float(input("Enter Employee Total working hours:"))
Hourly_rate=float(input("Enter Employee Hourly rate:"))

print("\nEmployee Name:",name)
normal_hoursalary=40*Hourly_rate

if T_workhours == 40:
    print("Normal Hours Salary:Rs",normal_hoursalary)

else:
    ex_workhours=T_workhours-40
    ex_hourrate=Hourly_rate*1.5
    ex_hoursalary=ex_workhours * ex_hourrate
    weekly_salary=normal_hoursalary + ex_hoursalary

    print("\nNormal Hours Salary:Rs",normal_hoursalary)
    print("\nExtra Work hours:", ex_workhours)
    print("\nExtra Hours Salary:Rs",ex_hoursalary)
    print("\nWeekly Salary:Rs",weekly_salary)
```

```python
#Task 9
# Message analyzer

user_msg=input("Enter your Message:")
Character=(len(user_msg))
print("Character:", Character)

Words=len(user_msg.split())
print("Words:",Words)

Space=user_msg.count(" ")
print("Space:",Space)

print("User Message in Uppercase:",user_msg.upper())

vowels=["a","e","i","o","u"]
vowel_count=0
for character in user_msg:
    if character in vowels:
        vowel_count=vowel_count+1
print("Vowels:",vowel_count)

if "Python" in user_msg:
    print("Python Found:","Yes")
else:
    print("Python Found:","No")
```
