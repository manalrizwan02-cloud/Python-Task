```python
#Task 11: Personal Expense Tracker
Total_expense=int(input("Enter your Total Expenses:"))

daily_expenses=[]

for i in range(1,Total_expense+1):
     expense_name=input("Enter your Expense Name:")
     category_name=input("Enter your Expense Category Name:").title()
     amount=int(input("Enter your Expense Amount:"))

     expense={
       "Name":expense_name,
       "Category":category_name,
       "Amount":amount
     }

     daily_expenses.append(expense)
print("Daily Expenses :", daily_expenses)

total_spending=0
for expense in daily_expenses:
    total_spending+=expense["Amount"]


Highest_expense=0

for expense in daily_expenses:
    print(expense["Category"],":",expense["Amount"])

    if expense["Amount"] > Highest_expense:
         Highest_expense= expense["Amount"]

category_spending = {
    "Food": 0,
    "Transport": 0,
    "Shopping":0,
    "Bill":0
}
for expense in daily_expenses:
    category = expense["Category"]
    amount = expense["Amount"]
    category_spending[category] += amount

budget=float(input("Enter your Budget:"))

if total_spending > budget:
    Budget_Status="Warning Budget Exceeded!"
else:
   Budget_Status= "You're in your Budget"

def final_report():
    print("\n====== Total Expense Report======")
    print("Total_spending:",total_spending)
    print("Category Spending:")
    for category, amount in category_spending.items():
        print(category,":",amount)
    print("Highest Expense:",Highest_expense)
    print("Budget Status:",Budget_Status)
    return
final_report()
```

```python
# Task 12: Store Inventory Manager

products = {}

def add_product():

    name = input("Enter Product Name: ").title()

    if name in products:
        print("Product already exists")
        return

    price = int(input("Enter Product Price: "))
    quantity = int(input("Enter Available Quantity: "))

    products[name] = {
        "price": price,
        "quantity": quantity
    }

    print("Product Added Successfully")

def sell_product():

    name = input("Enter Product Name: ").title()

    if name in products:

        quantity = int(input("Enter Quantity to Sell: "))

        if quantity <= products[name]["quantity"]:

            products[name]["quantity"] -= quantity

            print("Product Sold Successfully")
            print("Remaining Stock:", products[name]["quantity"])

        else:
            print("Insufficient Stock")

    else:
        print("Product Not Found")

def restock_product():

    name = input("Enter Product Name: ").title()

    if name in products:

        quantity = int(input("Enter Quantity to Add: "))

        products[name]["quantity"] += quantity

        print("Product Restocked Successfully")
        print("Updated Stock:", products[name]["quantity"])

    else:
        print("Product Not Found")

def search_product():

    name = input("Enter Product Name: ").title()

    if name in products:

        print("\nProduct Found:")
        print("Product Name:", name)
        print("Price:", products[name]["price"])
        print("Available Quantity:", products[name]["quantity"])

    else:
        print("Product Not Found")

def low_stock():

    print("\n------ Low Stock Products ------")

    for name, product in products.items():


        if product["quantity"] < 5:

            print("Product Name:", name)
            print("Price:", product["price"])
            print("Available Quantity:", product["quantity"])
            print()


    print("Low Stock product not found")

def inventory_value():

    total = 0

    for name, product in products.items():

        total = total + (product["price"] * product["quantity"])

    print("\nTotal Inventory Value:", total)

while True:

    print("\n===== Store Inventory Manager =====")
    print("1. Add Product")
    print("2. Sell Product")
    print("3. Restock Product")
    print("4. Search Product")
    print("5. Display Low-Stock Products")
    print("6. Total Inventory Value")
    print("7. Exit")

    choice = int(input("Enter Choice: "))

    if choice == 1:
        add_product()

    elif choice == 2:
        sell_product()

    elif choice == 3:
        restock_product()

    elif choice == 4:
        search_product()

    elif choice == 5:
        low_stock()

    elif choice == 6:
        inventory_value()

    elif choice == 7:
        print("Thank You!")
        break

    else:
        print("Invalid Choice")
```

```python
#Task 13:Library Book Management System

books = []

def add_book():

    id = int(input("Enter Book ID: "))

    for book in books:
        if book["ID"] == id:
            print("Book ID already exists")
            return

    title = input("Enter Book Title: ")
    author = input("Enter Author Name: ")
    status = input("Enter the Availability Status:")

    book = {
        "ID": id,
        "Title": title,
        "Author": author,
        "Availability": status
    }

    books.append(book)

    print("Book Added Successfully")

def search_book():

    title = input("Enter Book Title: ").title()

    for book in books:

        if book["Title"] == title:

            print("\nBook Found:")

            print("Book ID:", book["ID"])
            print("Book Title:", book["Title"])
            print("Author:", book["Author"])
            print("Availability:", book["Availability"])

            return

    print("Book Not Found")

def borrow_book():

    id = int(input("Enter Book ID: "))

    for book in books:

        if book["ID"] == id:

            if book["Availability"] == "Available":

                book["Availability"] = "Borrowed"

                print("Book Borrowed Successfully")

            else:

                print("Book is not available")

            return

    print("Book Not Found")

def return_book():

    id = int(input("Enter Book ID: "))

    for book in books:

        if book["ID"] == id:

            if book["Availability"] == "Borrowed":

                book["Availability"] = "Available"

                print("Book Returned Successfully")

            else:

                print("Book is already available")

            return

    print("Book Not Found")

def display_books():

    print("\n------ All Books ------")

    for book in books:

        print("Book ID:", book["ID"])
        print("Book Title:", book["Title"])
        print("Author:", book["Author"])
        print("Availability:", book["Availability"])
        print()


while True:

    print("\n===== Library Book Management System =====")
    print("1. Add Book")
    print("2. Search Book")
    print("3. Borrow Book")
    print("4. Return Book")
    print("5. Display All Books")
    print("6. Exit")

    choice = int(input("Enter Choice: "))

    if choice == 1:
        add_book()

    elif choice == 2:
        search_book()

    elif choice == 3:
        borrow_book()

    elif choice == 4:
        return_book()

    elif choice == 5:
        display_books()

    elif choice == 6:
        print("Thank You!")
        break

    else:
        print("Invalid Choice")
```
