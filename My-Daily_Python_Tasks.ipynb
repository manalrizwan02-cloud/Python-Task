```python
#product search and update
products = {
"Laptop": {"price": 80000, "quantity": 5},
"Mouse": {"price": 1500, "quantity": 10},
"Keyboard": {"price": 3000, "quantity": 4}
}
name=input("Enter the Product Name:")
if name in products:
    print("Product exist")
    user_quantity=int(input("Enter the Product Quantity:"))

    if user_quantity <=  products[name]["quantity"]:
        print("quantity available")
        update_quantity = products[name]["quantity"]-user_quantity
        products[name]["quantity"]=update_quantity
        print(update_quantity)
    else:

        print("quantity insufficient")
```

```python
#Function based Result
marks = int(input("Enter marks: "))
def calculate_result(marks):
    if marks >=80:
        print("A+ Grade")
    elif marks >=70:
        print("A Grade")
    elif marks >=60:
        print("B Grade")
    elif marks >=50:
        print("C Grade")
    elif marks > 40:
        print("D Grade")
    else:
        print("Fail")
calculate_result(marks)
```

```python
#Shopping Cart checker
products = {
    "apple": 100,
    "banana": 50,
    "mango": 200,
    "orange": 80
}
total_products=int(input("How many product do you want ?"))
total_bill = 0
for i in range(1,total_products +1):
    name=input("Enter the Product Name:")
    if name in products:
        print("Product Available")
        quantity= int(input("How much quantity do you want?"))
        prise=products[name]
        sub_total=prise*quantity
        total_bill=total_bill+sub_total
    else:
        print("Product Not available")
def Final_reciept():
      print("\n Final Reciept")
      print("Total Bill:",total_bill)
Final_reciept()
```

```python
#Student Analyzer
students = {
    "Ali": 78,
    "Sara": 92,
    "Ahmed": 65,
    "Hina": 88,
    "Noman": 45
}

total_students=len(students)
print("Total Students:",total_students)

pass_count=0
fail_count=0
highest_marks=0
highest_student=""
total_marks=0

for student,marks in students.items():
    total_marks+=marks

    if marks>=50:
        pass_count+=1
    else:
        fail_count+=1

    if marks > highest_marks:
        highest_marks=marks
        highest_student=student

average_marks= total_marks/total_students

print("Total Pass Students:",pass_count)
print("Total Fail Students:",fail_count)
print("Highest Student Marks=",highest_student,":",highest_marks)
print("Average Marks of Students:",average_marks)
```

```python
#Mini Bank Management
accounts = {
    "Ali": 5000,
    "Sara": 8000,
    "Ahmed": 3000
}

account_name=input("Enter Account Name:")
if account_name in accounts:
    print("Start the Process")
else:
    print("Invalid Account")

while True:
    print("----Operations----")
    print("1:Deposit")
    print("2: Withdrawal")
    print("3:Exit")


    choice=int(input("Enter the choice you want:"))
    if choice == 1:
        amount=float(input("Enter the amount for deposit:"))
        current_amount=accounts[account_name]
        deposit_increase=current_amount+amount
        accounts[account_name]=deposit_increase
        print("\nUpdated Account:",accounts[account_name])

    elif choice == 2:
        amount = float(input("Enter the amount for withdrawal:"))

        current_amount = accounts[account_name]

        if amount <= current_amount:
            new_balance = current_amount - amount
            accounts[account_name] = new_balance
            print("\nUpdated Balance:", accounts[account_name])
        else:
            print("Account Balance is insufficient")

    elif choice == 3:
        print("Thank you")
        break

    else:
        print("Invalid Choice")
```

```python
#word frequency
user_msg=input("Enter the Message:")
Count=len(user_msg.split())
print("Count:", Count)

frequency={}
for word in user_msg.split():
  if word in frequency:
      frequency[word]+=1
  else:
      frequency[word]=1
for word,Count in frequency.items():

  print("\n",word,":",Count)
```

```python
#Employee salary analyzer
employees = {
    "Ali": 45000,
    "Sara": 70000,
    "Ahmed": 55000,
    "Hina": 80000,
    "Noman": 40000
}
count=len(employees)
print("Total Employee:",count)

total_salary=0
average_salary=0

for salary in employees.values():
    total_salary+=salary
    average_salary=total_salary/count

print("Total Salary:",total_salary)
print("Average Salary:",average_salary)

highest_salary =max(employees.values())
lowest_salary=min(employees.values())

highest_employee=""
lowest_employee=""

for employee,salary in employees.items():
    if salary == highest_salary:
        highest_employee = employee
    if salary == lowest_salary:
        lowest_employee = employee

print("\nHighest Employee Salary = ",highest_employee,":",highest_salary)
print("\nLowest Employee Salary=",lowest_employee,":",lowest_salary)

employee =input("Enter the Employee Name:")

if employee in employees:
    print("Employee Exist")
    salary=employees[employee]
    print(employee,":",salary)
else:
    print("Employee Not Found")
```

```python
#Product Sale Analyzer
sales = {
    "Laptop": 5,
    "Mouse": 12,
    "Keyboard": 8,
    "Monitor": 3,
    "Headphones": 10
}

total_products = len(sales)
print("Total Products:",total_products)

t_unitssold = 0
average_unitsold = 0

for unit_sold in sales.values():
    t_unitssold += unit_sold
    average_unitsold = t_unitssold / total_products

print("\nTotal Units Sold:",t_unitssold)
print("\naverage_unitsold:",average_unitsold)

highest_saleproduct = max(sales.values())
lowest_saleproduct = min(sales.values())

print("\nHighest Selling Product:",highest_saleproduct)
print("\nLowest Selling Product:",lowest_saleproduct)

highest_product =""
lowest_product =""

for product,sale in sales.items():
    if sale == highest_saleproduct:
        highest_product = product
    if sale == lowest_saleproduct:
        lowest_product = product
print("\nHighest Selling Product:",highest_product,":",highest_saleproduct)
print("\nLowest Selling Product:",lowest_product,":",lowest_saleproduct)

product_name = input("Enter the Product Name from Sale:")
if product_name in sales:
    print("Product Exist")
    sale = sales[product_name]

    print(product_name,":",sale)
else:
    print("Product Not Exist")
```
