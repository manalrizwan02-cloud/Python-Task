```python
#Question2
x=float(input("Enter the number:"))
y=float(input("Enter the number:"))
operator=input("Enter the operator(+,-,*,/):")
if operator=="+":
    print(x+y)
elif operator=="-":
    print(x-y)
elif operator=="*":
    print(x*y)
elif operator=="/":
    print(x/y)
else:
    print("Invalid operator")
```
