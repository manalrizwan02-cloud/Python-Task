```python
Total_expense=int(input("Enter your Total Expenses:"))

daily_expenses=[]

for i in range(1,Total_expense+1):
     expense_name=input("Enter your Expense Name:")
     category_name=input("Enter your Expense Category Name:").title()
     amount=int(input("Enter your Expense Amount:"))

     expense={
       "Name":expense_name,
       "Category":category_name,
       "Amount":amount
     }

     daily_expenses.append(expense)
print("Daily Expenses :", daily_expenses)

total_spending=0
for expense in daily_expenses:
    total_spending+=expense["Amount"]


Highest_expense=0

for expense in daily_expenses:
    print(expense["Category"],":",expense["Amount"])

    if expense["Amount"] > Highest_expense:
         Highest_expense= expense["Amount"]

category_spending = {
    "Food": 0,
    "Transport": 0,
    "Shopping":0,
    "Bill":0
}
for expense in daily_expenses:
    category = expense["Category"]
    amount = expense["Amount"]
    category_spending[category] += amount

budget=float(input("Enter your Budget:"))

if total_spending > budget:
    Budget_Status="Warning Budget Exceeded!"
else:
   Budget_Status= "You're in your Budget"

def final_report():
    print("\n====== Total Expense Report======")
    print("Total_spending:",total_spending)
    print("Category Spending:")
    for category, amount in category_spending.items():
        print(category,":",amount)
    print("Highest Expense:",Highest_expense)
    print("Budget Status:",Budget_Status)
    return
final_report()
```

```python
# Store Inventory System

products_list = []


# 1. Add Product
def Add_Product():
    name = input("Enter Product Name: ")
    price = int(input("Enter Product Price: "))
    quantity = int(input("Enter Available Quantity: "))

    product = {
        "name": name,
        "price": price,
        "quantity": quantity
    }

    products_list.append(product)

    print("Product added successfully.")


# 2. Sell Product
def Sell_Product():
    name = input("Enter Product Name: ")

    for product in products_list:

        if name == product["name"]:
            print("Product Found!")

            quantity = int(input("Enter Quantity to Sell: "))

            if quantity <= product["quantity"]:
                product["quantity"] -= quantity
                print("Product sold successfully.")
            else:
                print("Insufficient Quantity.")

            break

    else:
        print("Product not found.")


# 3. Restock Product
def Restock_Product():
    name = input("Enter Product Name: ")

    for product in products_list:

        if name == product["name"]:
            print("Product Found!")

            quantity = int(input("Enter Quantity to Restock: "))

            product["quantity"] += quantity

            print("Product restocked successfully.")

            break

    else:
        print("Product not found.")


# 4. Search Product
def Search_Product():
    name = input("Enter Product Name: ")

    for product in products_list:

        if name == product["name"]:
            print("Product Found!")
            print("Price:", product["price"])
            print("Quantity:", product["quantity"])

            break

    else:
        print("Product not found.")


# 5. Display Inventory
def Display_Inventory():

    total_value = 0

    print("\n===== Inventory =====")

    for product in products_list:

        print("\nProduct Name:", product["name"])
        print("Price:", product["price"])
        print("Quantity:", product["quantity"])

        product_value = product["price"] * product["quantity"]

        print("Product Value:", product_value)

        total_value += product_value

        if product["quantity"] < 5:
            print("Low Stock!")

    print("\nTotal Inventory Value:", total_value)


# Main Menu
while True:

    print("\n===== Store Inventory Menu =====")
    print("1. Add Product")
    print("2. Sell Product")
    print("3. Restock Product")
    print("4. Search Product")
    print("5. Display Inventory")
    print("6. Exit")

    choice = input("Enter your choice: ")

    if choice == "1":
        Add_Product()

    elif choice == "2":
        Sell_Product()

    elif choice == "3":
        Restock_Product()

    elif choice == "4":
        Search_Product()

    elif choice == "5":
        Display_Inventory()

    elif choice == "6":
        print("Program ended.")
        break

    else:
        print("Invalid choice.")
```
